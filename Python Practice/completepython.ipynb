```python
### Prime numbers in between 1 to 100
for num in range (1,101):
    is_prime = True

    if num >1:
        for i in range (2, num):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            print(num)
```

```python
num = 2

while num <= 100:
    is_prime = True
    i = 2

    while i < num:
        if num % i == 0:
            is_prime = False
            break

        i = i + 1

    if is_prime:
        print(num)

    num = num+1

    
```
