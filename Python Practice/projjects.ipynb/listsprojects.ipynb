```python
## manage a to do list
to_do_list = ["complete classes","eat properly","complete assignments"]

## add new tasks 
to_do_list.append("room cleaning")
to_do_list.append("complete exersize")

## remove a completed task
to_do_list.remove("complete classes")

## cheching if a task in the list
if "complete assignments" in to_do_list:
    print("compllete your assignment properly")


print("to_do_list remaining")
for tasks in to_do_list:
    print(f"-{tasks}")

```

```python
## create a grade list for a student 
grades = [78,90,89,53,94]

## add a grade 
grades.append(84)

## calculate the average grade
avg_grade = sum(grades) / len(grades)
print(f"average grade :{avg_grade:.2f}")

## finding the highest and the lowest grade 
highest_grade = max(grades)
lowest_grade = min(grades)

print(f"highrst grade is : {highest_grade}")
print(f"lowest grade is : {lowest_grade}")
```
